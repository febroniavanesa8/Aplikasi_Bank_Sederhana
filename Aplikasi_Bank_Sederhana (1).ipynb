```python
class Rekening:

    def __init__(self, no_rekening, nama, saldo):
        self.no_rekening = no_rekening
        self.nama = nama
        self.saldo = saldo

    def setor(self, jumlah):
        self.saldo += jumlah
        print("Setoran berhasil")

    def tarik(self, jumlah):
        if jumlah <= self.saldo:
            self.saldo -= jumlah
            print("Penarikan berhasil")
        else:
            print("Saldo tidak mencukupi")

    def tampil_saldo(self):
        print("Saldo :", self.saldo)


rekening = {}

while True:

    print("\n===== APLIKASI BANK =====")
    print("1. Buat Rekening")
    print("2. Setoran")
    print("3. Penarikan")
    print("4. Cek Saldo")
    print("5. Data Nasabah")
    print("6. Keluar")

    pilih = input("Pilih Menu : ")

    if pilih == "1":

        no = input("Nomor Rekening : ")
        nama = input("Nama Nasabah : ")
        saldo = float(input("Saldo Awal : "))

        rekening[no] = Rekening(no, nama, saldo)

        print("Rekening berhasil dibuat")

    elif pilih == "2":

        no = input("Nomor Rekening : ")

        if no in rekening:
            jumlah = float(input("Jumlah Setoran : "))
            rekening[no].setor(jumlah)
            print("Saldo sekarang :", rekening[no].saldo)
        else:
          print("Rekening tidak ditemukan")

    elif pilih == "3":

        no = input("Nomor Rekening : ")

        if no in rekening:
            jumlah = float(input("Jumlah Penarikan : "))
            rekening[no].tarik(jumlah)
            print("Saldo sekarang :", rekening[no].saldo)
        else:
            print("Rekening tidak ditemukan")
    elif pilih == "4":

        no = input("Nomor Rekening : ")

        if no in rekening:
            rekening[no].tampil_saldo()

    elif pilih == "5":

        for r in rekening.values():

            print("No Rekening :", r.no_rekening)
            print("Nama :", r.nama)
            print("Saldo :", r.saldo)

    elif pilih == "6":
        break
```
